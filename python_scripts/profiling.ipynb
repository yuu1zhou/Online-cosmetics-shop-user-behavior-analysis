```python
import pandas as pd
import numpy as np
import os
```

```python
parent_dir = os.path.dirname(os.getcwd())
years = [2019, 2020]
df = pd.DataFrame()

for year in years:
    if year == 2019:
        months = ['Oct', 'Nov', 'Dec']
        for month in months:
            df_temp = pd.read_csv(os.path.join(parent_dir, f'./{year}-{month}.csv'))
            df = pd.concat([df, df_temp], axis = 0)
    else:
        months = ['Jan', 'Feb']
        for month in months:
            df_temp = pd.read_csv(os.path.join(parent_dir, f'./{year}-{month}.csv'))
            df = pd.concat([df, df_temp], axis=0)
```

# NA rate

```python
for col in df.columns:
    print(f'{col} na rate: ', np.sum(df[col].isna()) * 100 / df.shape[0])
```

# Column unique values

categorical: `event_type`, `category_code`, `brand`

```python
for col in df.columns:
    print(f'{col} unqiue values: ', len(df[col].value_counts()))
```

```python
for col in ['event_type', 'category_id', 'category_code', 'brand']:
    print(col + ':', list(df[col].value_counts().index))
```
